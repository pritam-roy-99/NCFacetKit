# NCFacetKit v0.2 research tutorial
A finite generalized contextuality scenario from symbolic equations to facets, exact noncontextual bounds, and observed contextual fraction.


```python
from ncfacetkit import Scenario
s = Scenario(4, 2, 2).prepare("(P0+P1)/2=(P2+P3)/2")
print(s.analyze())
```

```python
result=s.facets()
result.show(30)
print("Completeness:", result.completeness)
print("Orbit sizes:", [k for _,k in s.orbits(result)])
```

```python
score = "p(0|1,0)-p(0|2,0)-p(0|0,1)+p(0|2,1)"
b=s.noncontextual_bound(score)
print(b)
print("Witness:", b.probabilities)
```
